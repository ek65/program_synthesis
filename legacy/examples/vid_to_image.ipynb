```python
import cv2
import os
import numpy as np

def extract_frames(video_path, n_frames, output_folder="frames"):
    # Create the output folder if it doesn't exist
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)
    
    # Open the video file
    cap = cv2.VideoCapture(video_path)
    if not cap.isOpened():
        print(f"Error opening video file: {video_path}")
        return

    # Get total number of frames in the video
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    if total_frames == 0:
        print("No frames found in video!")
        return

    # Generate n evenly spaced frame indices from 0 to total_frames - 1
    frame_indices = np.linspace(0, total_frames - 1, n_frames, dtype=int)

    for i, frame_index in enumerate(frame_indices):
        # Set the video position to the specific frame index
        cap.set(cv2.CAP_PROP_POS_FRAMES, frame_index)
        ret, frame = cap.read()
        if not ret:
            print(f"Warning: Could not read frame at index {frame_index}")
            continue
        
        # Save the frame to the output folder
        frame_filename = os.path.join(output_folder, f"frame{i}.jpg")
        cv2.imwrite(frame_filename, frame)
        print(f"Saved {frame_filename}")

    cap.release()

if __name__ == "__main__":
    video_path = "envelope2.mp4"  # Replace with your video file path
    n_frames = 168  # Number of frames to extract (including the first and last frame)
    extract_frames(video_path, n_frames)
```
