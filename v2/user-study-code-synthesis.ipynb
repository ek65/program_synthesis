```python
import os
from unity_utils import UnityTranslator
import json
from scenic_fc.api import api
from infer_utils import Inference

user_name = 'user1'

dir = os.getcwd()
DATA_DIR = dir + '/data/real-give-and-go2-1'
demos = UnityTranslator.get_from(DATA_DIR)
acts = Inference.act_from(demos, api, model = 'o3-mini')
combined = Inference.combine(acts = acts, api = api, model = 'o3-mini')

import json
def save_json(file_name, act):
    OUT_DIR = 'exports'
    
    os.makedirs(OUT_DIR, exist_ok=True)
    data = act.to_dict(id=file_name, expanded=True)
    
    with open(f'{OUT_DIR}/{file_name}.json', 'w') as f:
        json.dump(data, f, indent=4)
```
